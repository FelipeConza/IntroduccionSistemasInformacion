```python
#Análisis de supervivencia

print("El porcentaje de pasajeros que sobrevivieron: ", sum(df.Survived)/len(df)*100, "%")
```

```python
#Análisis de clase
print("El porcentaje de pasajeros de primera clase fue: ", df['Pclass'].value_counts()[1]/len(df)*100, "%")
print("El porcentaje de pasajeros de segunda clase fue: ", df['Pclass'].value_counts()[2]/len(df)*100, "%")
print("El porcentaje de pasajeros de tercera clase fue: ", df['Pclass'].value_counts()[3]/len(df)*100, "%")
```

```python
#Análisis de género
print("El número de hombres que abordaron el Titanic fue: ", df['Sex'].value_counts()['male'])
print("El número de mujeres que abordaron el Titanic fue: ", df['Sex'].value_counts()['female'])
```
